# Part 1

```python
answer = 0
for id_range in input_file:

    start,end = map(lambda x : int(x), id_range.split('-'))
    for value in range(start,end + 1):
        str_value = str(value)
        if len(str_value)%2 ==0:
            half = len(str_value)// 2
            left = int(str_value[:half])
            right = int(str_value[half:])
            if left == right:
                answer = answer + value
print(answer)
```

# Part 2

```python
def chunk(value,increase):
    aux = []
    for i in range(0,len(value),increase):
        aux.append(value[i:i+increase])
    return aux

def repeat_hunter(value):
    len_value = len(value)
    if len_value == 2:
        return value[0] == value[1]
    elif len_value == 3:
        return len(set(value)) == 1
    else:
        if len(set(value)) == 1:
            return True
        list_factor = divisors(int(len_value))[1:-1]
        for factor in list_factor:
            chunk_list = chunk(value,factor)
            if len(set(chunk_list)) == 1:
                return True
        
```

```python
answer = 0
for id_range in input_file:
    start,end = map(lambda x : int(x), id_range.split('-'))
    for value in range(start,end + 1):
        str_value = str(value)
        if repeat_hunter(str_value):
            answer = answer + value
print(answer)
#correct answer 24774350322 
```
